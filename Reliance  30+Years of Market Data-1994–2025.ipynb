```python
Top_5Days_high_Tornover = np.array([1.473430e+11, 1.236080e+11, 	1.184670e+11, 9.211999e+10, 9.179980e+10])
Top_5Daays_low_Tornover = np.array([1.068075e+06, 2.288825e+06, 2.515898e+06, 2.515898e+06, 2.598205e+06])
Top_days = np.array(['10-09-2020', '15-07-2020', '24-07-2020', '01-07-2022','24-06-2021'])
Low_days = np.array(['29-12-1994', '28-11-1994', '27-12-1994', '27-12-1994', '28-12-1994'])
plt.plot(Top_days, Top_5Days_high_Tornover, label='Tornover')
plt.title('Top 5 High profit days in 30 Years')
plt.xlabel('Days/Dates')
plt.ylabel('Tornover')
plt.legend()
plt.show()
```

```python
plt.plot(Low_days, Top_5Daays_low_Tornover, label='Tornover') 
plt.title('Top 5 low profit days in 30 Years')
plt.legend()
plt.xlabel('Days/Dates')
plt.ylabel('Tornover')
plt.show()
```
